```````python
```# Hypothesis: There is no significant difference in median value of houses bounded by the Charles river or not.
# H0: μ1 = μ2
# Ha: μ1 ≠ μ2

bounded = boston_df[boston_df['CHAS'] == 1]['MEDV']
not_bounded = boston_df[boston_df['CHAS'] == 0]['MEDV']

t_stat, p_value = stats.ttest_ind(bounded, not_bounded)
print(f'T-statistic: {t_stat}, P-value: {p_value}')

if p_value < 0.05:
    print('Reject the null hypothesis: There is a significant difference.')
else:
    print('Fail to reject the null hypothesis: No significant difference.')```
```````

```````python
```# Hypothesis: There is no difference in median values of houses for each proportion of owner-occupied units built prior to 1940.
# H0: μ1 = μ2 = μ3
# Ha: At least one μ is different.

model = ols('MEDV ~ C(AGE_GROUP)', data=boston_df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)

if anova_table['PR(>F)'][0] < 0.05:
    print('Reject the null hypothesis: There is a significant difference.')
else:
    print('Fail to reject the null hypothesis: No significant difference.')```
```````

```````python
```# Hypothesis: There is no impact of an additional weighted distance to the five Boston employment centres on the median value of owner-occupied homes.
# H0: β1 = 0
# Ha: β1 ≠ 0

X = boston_df['DIS']
y = boston_df['MEDV']
X = sm.add_constant(X)  # Adds a constant term to the predictor

model = sm.OLS(y, X).fit()
print(model.summary())

if model.pvalues[1] < 0.05:
    print('Reject the null hypothesis: There is a significant impact.')
else:
    print('Fail to reject the null hypothesis: No significant impact.')```
```````
